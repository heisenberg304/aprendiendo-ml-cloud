```python
#importar librerías
import numpy as np

# Entradas
x = np.array([2, 3, 1])

# Pesos
w = np.array([0.5, -0.2, 0.8])

#bias
b = 0.1


#funciones de activacion
#Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
#Relu
def relu(x):
    return np.maximum(0, x)
#relu(z)
#Tanh
def tanh(x):
    return np.tanh(x)
#tanh(z)

#prediccion
z = np.dot(x, w) + b
pred = sigmoid(z)

print(pred)
```
