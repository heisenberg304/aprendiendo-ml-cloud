```python
import numpy as np
import matplotlib.pyplot as plt

#dataset
x = np.array([90, 900, 1000, 1100, 1200, 1300, 1450, 1600, 1750, 1900, 2100, 4000]) #peso en kg
y = np.array([0.48, 4.8, 5.2, 5.7, 6.1, 6.6, 7.2, 7.8, 8.4, 9.1, 9.9, 20]) #consumo en L/1000mk

#valores
w = 1
b = 1
a = 0.00000031

#gradiente
def loss(w, b):
  #prediccion
  y_hat = x*w + b

  #error
  error = y_hat - y

  #funcion de error
  L = sum((error)**2)/len(x)

  #derivada de mse respecto a w and b
  Lw = sum(2*x*(error))/len(x)
  Lb = sum(error)/len(x)
  return (Lw, Lb, L)

def graficar(w,b):
  y_hat = x*w + b
  plt.scatter(x, y_hat)
  plt.xlabel("x")
  plt.ylabel("y")
  plt.title('relacion entre x and y')
  return plt.show()

for epoch in range(1000):
  W = w
  B = b
  grad_w, grad_b, L = loss(w, b)
  w = w - a*(grad_w)
  b = b - a*(grad_b)
  if epoch % 100 == 0:
    graficar(w, b)

  print(f"iter {epoch+1}, {W} ----> {w}             {B} ----> {b}")
  print(f"{L}")

```
