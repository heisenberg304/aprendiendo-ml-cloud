```python
#importar librerias
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
import pandas as pd

#CARGAR DATOS
data = pd.read_csv("titanic.csv")
#data.head()   #primeras filas
#data.columns  #ver columnas
#data.info()   #informacion general
data.isnull().sum() #cuántos datos faltan por columna

#Aplicar mediana para rellenar
data["Age"].fillna(data["Age"].median(), inplace=True)
#data["Age"] ver todas las edades
#.fillna() rellena
#median() hace automaticamente la mediana
#inplace=True, si el sitio esta vacio, osea True, es rellenado
data.isnull().sum() #ahora despues de rellenar con mediana (mas robusta), no le falta ningun dato en Age

#encoding one-hot
fe_ma = pd.get_dummies(data["Sex"]) #toma sex: male, female y lo convierte en sex: [1,0], [0,1], osea en vaiables binarias solo con un 1

#unir las nuevas filas al dataframe principal
data = pd.concat([data, fe_ma], axis=1)  #pegar dos dataframes uno al lado del otro

#eliminar linea sobrantes:
data = data.drop(["Sex", "PassengerId", "Name", "SibSp", "Parch", "Ticket", "Cabin", "Embarked"], axis=1)  #para eliminar columnas

#Separar x and y
x = data[["Age", "Pclass", "female", "male", "Fare"]]
y = data["Survived"]

#Separar dataset en train y test
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size = 0.2,
    random_state = 42
)

#crear modelo
model = LogisticRegression()
#entrenar el modelo
model.fit(x_train, y_train)
#testear modelo
pred = model.predict(x_test)

#coeficientes e intercepto
w = model.coef_
b = model.intercept_

#metricas
matrix = confusion_matrix(y_test, pred)
score = accuracy_score(y_test, pred)
precision = precision_score(y_test, pred)
recall = recall_score(y_test, pred)
f1 = f1_score(y_test, pred)

print(pred)
print(y_test)
print(f"coeficiente w = {w}, intercepto = {b}")
print(f"matrix -> {matrix}")
print(f"score -> {score}")
print(f"precision -> {precision}")
print(f"recall -> {recall}")
print(f"f1 score -> {f1}")
#print(x_train.shape) #(712, 5)
#print(x_test.shape)  #(179, 5)
```
