```python
import numpy as np
#vectores
x = np.array([1,2,3,4])
y = np.array([2,3,5,4])

#parametros: w, b and a
w = 3
b = 2
a = 0.1

#prediccion
y_hat = w*x + b

#error
error = y_hat - y

#funcion error
mse = (sum(error**2))/len(x)

#derivada de funcion error respecto a w and b
def grad_w_b(w,b):
  y_hat1 = w * x + b
  error = y_hat1 - y
  return ((sum(2*x*(error)))/len(x), (sum(error))/len(x))

#descenso de gradiente de w and b
for i in range(10):
  b1 = b
  w1 = w
  grad_w, grad_b = grad_w_b(w,b)
  b = b - a * grad_b
  w = w - a * grad_w
  print(f"iter {i}, {w1} -> {w}  ---- iter {i}, {b1} -> {b} \n")
```
