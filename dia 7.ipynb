```python
#importar librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score

#importar datos
data = pd.read_csv("/content/drive/MyDrive/data_csv/Give_Me_Some_Credit.csv")

#mi analisis en resumen:
#print((data["MonthlyIncome"]<=6670).value_counts()) #atravez de un analisis, conclui que muchas personas ganan mas que la mediana y promedio, y al haber 150k de casos me puedo dar el lujo de eliminar estas filas

#eliminar datos faltantes
data.dropna(subset="MonthlyIncome", inplace=True) #esto elimina tambien los faltantes de la otra columna "NumberOfDependents"

#definir parametros
x = data[["RevolvingUtilizationOfUnsecuredLines",
         "age",
         "NumberOfTime30-59DaysPastDueNotWorse",
         "DebtRatio",
         "MonthlyIncome",
         "NumberOfOpenCreditLinesAndLoans",
         "NumberOfTimes90DaysLate",
         "NumberRealEstateLoansOrLines",
         "NumberOfTime60-89DaysPastDueNotWorse",
         "NumberOfDependents"]]
y = data["SeriousDlqin2yrs"]
#Split
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size = 0.2,
    random_state = 42
)

#crear modelo
model = DecisionTreeClassifier(
    class_weight = {0:1, 1:9}, #para lidiar con desbalance
    max_depth=9
)

#entrenar modelo
model.fit(x_train, y_train)

#prediccion - threshold
#train
probs_train = model.predict_proba(x_train)[:, 1]
threshold = 0.4
pred_train = (probs_train >= threshold).astype(int)
#test
probs = model.predict_proba(x_test)[:, 1]
threshold = 0.4
pred = (probs >= threshold).astype(int)

#metricas
matrix = confusion_matrix(y_test, pred)
accuracy = accuracy_score(y_test, pred)
train_score = accuracy_score(y_train, pred_train)
precision = precision_score(y_test, pred)
recall = recall_score(y_test, pred)
f1 = f1_score(y_test, pred)

print(f"matrix -> {matrix}")
print(f"accuracy -> {accuracy}")
print(f"precision -> {precision}")
print(f"recall -> {recall}")
print(f"f1 score -> {f1}")
print(f"train score -> {train_score}")
```

```python
#importar librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import learning_curve

#importar datos
data = pd.read_csv("/content/drive/MyDrive/data_csv/Give_Me_Some_Credit.csv")

#mi analisis en resumen:
#print((data["MonthlyIncome"]<=6670).value_counts()) #atravez de un analisis, conclui que muchas personas ganan mas que la mediana y promedio, y al haber 150k de casos me puedo dar el lujo de eliminar estas filas faltantes

#eliminar datos faltantes
data.dropna(subset="MonthlyIncome", inplace=True) #esto elimina tambien los faltantes de la otra columna "NumberOfDependents"

#definir parametros
x = data[["RevolvingUtilizationOfUnsecuredLines",
         "age",
         "NumberOfTime30-59DaysPastDueNotWorse",
         "DebtRatio",
         "MonthlyIncome",
         "NumberOfOpenCreditLinesAndLoans",
         "NumberOfTimes90DaysLate",
         "NumberRealEstateLoansOrLines",
         "NumberOfTime60-89DaysPastDueNotWorse",
         "NumberOfDependents"]]
y = data["SeriousDlqin2yrs"]
#Split
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size = 0.2,
    random_state = 42
)

#crear modelo
model = RandomForestClassifier(
    class_weight = {0:1, 1:8}, #para lidiar con desbalance
    n_estimators=200,
    max_depth=8,
    min_samples_leaf=15,
    min_samples_split=15,
    max_features="sqrt",
    random_state=1
)

#entrenar modelo
model.fit(x_train, y_train)

#prediccion - threshold
#train
probs_train = model.predict_proba(x_train)[:, 1]
threshold = 0.478
pred_train = (probs_train >= threshold).astype(int)
#test
probs = model.predict_proba(x_test)[:, 1]
threshold = 0.478
pred = (probs >= threshold).astype(int)

#metricas
matrix = confusion_matrix(y_test, pred)
accuracy = accuracy_score(y_test, pred)
train_score = accuracy_score(y_train, pred_train)
precision = precision_score(y_test, pred)
recall = recall_score(y_test, pred)
f1 = f1_score(y_test, pred)

print(f"matrix -> {matrix}")
print(f"accuracy -> {accuracy}")
print(f"precision -> {precision}")
print(f"recall -> {recall}")
print(f"f1 score -> {f1}")
print(f"train score -> {train_score}")


```

```python

#validacion cruzada
scores = cross_val_score(
    model,
    x,
    y,
    cv=5
)
print(f"""modelo:
mean -> {scores.mean()}
std -> {scores.std()}
train score -> {train_score}
gap -> {train_score - scores.mean()}""")


```
