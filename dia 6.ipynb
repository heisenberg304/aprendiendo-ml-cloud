```python
#importar librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler

#==========================================================================

#importar dataset
data = pd.read_csv("/content/drive/MyDrive/data_csv/house_prices.csv")

#==========================================================================

scaler = StandardScaler()
x1 = data[["size"]]
x2 = data[["rooms"]]
x3 = data[["age"]]
x4 = data[["distance_to_city"]]

data["size"] = scaler.fit_transform(x1)
data["rooms"] = scaler.fit_transform(x2)
data["age"] = scaler.fit_transform(x3)
data["distance_to_city"] = scaler.fit_transform(x4)

#==========================================================================
#obtener x and y
y = data["price"]
data = data.drop("price", axis=1)
x = data
#split al dataset
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=40
)

#==========================================================================

#regularizacion
ridge_1 = Ridge(alpha=0.1)
ridge_2 = Ridge(alpha=1)
ridge_3 = Ridge(alpha=10)
ridge_4 = Ridge(alpha=100)
#crear modelo
model = LinearRegression()
#entrenar modelos
model.fit(x_train, y_train)
ridge_1.fit(x_train, y_train)
ridge_2.fit(x_train, y_train)
ridge_3.fit(x_train, y_train)
ridge_4.fit(x_train, y_train)

#pred
pred = model.predict(x_test)
y_pred_1 = ridge_1.predict(x_test)
y_pred_2 = ridge_2.predict(x_test)
y_pred_3 = ridge_3.predict(x_test)
y_pred_4 = ridge_4.predict(x_test)

#coeficientes e interceptos
w = model.coef_
b = model.intercept_
#alpha = 0.1
L2w_1 = ridge_1.coef_
L2b_1 = ridge_1.intercept_
#alpha = 1
L2w_2 = ridge_2.coef_
L2b_2 = ridge_2.intercept_
#alpha = 10
L2w_3 = ridge_3.coef_
L2b_3 = ridge_3.intercept_
#alpha = 100
L2w_4 = ridge_4.coef_
L2b_4 = ridge_4.intercept_

#==========================================================================

#metricas para modelo base
r2 = r2_score(y_test, pred)
MSE = mean_squared_error(y_test, pred)
print(f"MSE --> {MSE}")
print(f"r2 --> {r2}")
print(f"coeficiente w = {w}, intercepto = {b}")

#metricas para modelo L2
#alpha = 0.1
L2_r2_1 = r2_score(y_test, y_pred_1)
L2_MSE_1 = mean_squared_error(y_test, y_pred_1)
#alpha = 1
L2_r2_2 = r2_score(y_test, y_pred_2)
L2_MSE_2 = mean_squared_error(y_test, y_pred_2)
#alpha = 10
L2_r2_3 = r2_score(y_test, y_pred_3)
L2_MSE_3 = mean_squared_error(y_test, y_pred_3)
#alpha = 100
L2_r2_4 = r2_score(y_test, y_pred_4)
L2_MSE_4 = mean_squared_error(y_test, y_pred_4)

#==========================================================================

results = pd.DataFrame({
    "Model": ["Linear", "Ridge 0.1", "Ridge 1", "Ridge 10", "Ridge 100"],
    "Alpha": [0, 0.1, 1, 10, 100],
    "MSE": [MSE, L2_MSE_1 ,L2_MSE_2 ,L2_MSE_3 ,L2_MSE_4],
    "R2": [r2, L2_r2_1, L2_r2_2, L2_r2_3, L2_r2_4],
    "Coefficients Behavior": [w, L2w_1, L2w_2, L2w_3, L2w_4],
    "Intercepto": [b, L2b_1, L2b_2, L2b_3, L2b_4]
})
results.sort_values(by="MSE")
display(results)
```
