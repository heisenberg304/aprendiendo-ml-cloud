```python
#importar librerías
import numpy as np

# Entradas
x = np.array([
    [2, 3, 1],
    [4, 2, 3],
    [1, 5, 2]
])
y = np.array([1, 0, 1])

# Pesos
w = np.array([0.5, -0.2, 0.8])

#bias
b = 0.1
learning_rate = 0.1

#funciones de activacion
#Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def grad_sigmoid(x):
  return sigmoid(x) * (1 - sigmoid(x))

#Relu
def relu(x):
    return np.maximum(0, x)
#relu(z)
#Tanh
def tanh(x):
    return np.tanh(x)
#tanh(z)

def f_loss(w,b):
  #prediccion
  z = np.dot(x, w) + b
  pred = sigmoid(z)
  #Error
  error = pred - y
  #MSE
  L = np.mean(error ** 2) / 2
  #dL/dz = error * sigmoid'(z)
  # Gradientes
  grad_w = np.mean(x * error[:, np.newaxis] * grad_sigmoid(z)[:, np.newaxis], axis=0)
  grad_b = np.mean(error * grad_sigmoid(z))

  return grad_w, grad_b, L, pred

for epoch in range(1000):
  W = w
  B = b
  grad_w, grad_b, L, pred = f_loss(w,b)
  w = w - (learning_rate*grad_w)
  b = b - (learning_rate*grad_b)
  if epoch % 100 == 0:
        print(f"Epoch: {epoch}")
        print(f"Loss: {L}")
        print(f"Predicción: {pred}")
        print(f"Pesos: {w}")
        print(f"Bias: {b}")
        print()
```
