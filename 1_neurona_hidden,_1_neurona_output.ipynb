```python
#importar librerías
import numpy as np

# Entradas
x = np.array([2])
y = np.array([6])

# Pesos
wo = np.array([0.5])
wh = np.array([0.7])

#bias
bo = np.array([0.1])
bh = np.array([0.2])

learning_rate = 0.1

#funciones de activacion
#Sigmoid
def sigmoid(x):
  return 1 / (1 + np.exp(-x))
def grad_sigmoid(x):
  return sigmoid(x) * (1 - sigmoid(x))

def predict(x,w,b):
  pred = np.dot(x, w) + b
  return pred

def error(pred, y):
  error = pred - y
  return error

def f_loss(error):
  #MSE
  L = np.mean(error ** 2) / 2
  return L

def actualizar(lr, w_b, grad):
  new = w_b - (lr*grad)
  return new

#no sabia como ponerle de nombre
def cosa(a=1, b=1, c=1, d=1, e=1):
  grad = a * b * c * d * e
  return grad

def main(x, wo, bo, wh, bh, y, lr):
  #hidden
  zh = predict(x, wh, bh)
  h = sigmoid(zh)
  #output
  zo = predict(h, wo, bo)
  output = sigmoid(zo)

  funcion_loss = f_loss(error(output, y))
  grad_wo = cosa(error(output, y), grad_sigmoid(zo), h)
  grad_bo = cosa(error(output, y), grad_sigmoid(zo))
  grad_wh = cosa(error(output, y), grad_sigmoid(zo), wo, grad_sigmoid(zh), x)
  grad_bh = cosa(error(output, y), grad_sigmoid(zo), wo, grad_sigmoid(zh))
  wo_new = actualizar(learning_rate, wo, grad_wo)
  bo_new = actualizar(learning_rate, bo, grad_bo)
  wh_new = actualizar(learning_rate, wh, grad_wh)
  bh_new = actualizar(learning_rate, bh, grad_bh)
  return output, funcion_loss, wo_new, bo_new, wh_new, bh_new

for epoch in range(1000):
  Wo = wo
  Bo = bo
  Wh = wh
  Bh = bh
  output, funcion_loss, wo, bo, wh, bh = main(x, wo, bo, wh, bh, y, learning_rate)
  #print(output, funcion_loss, wo, bo, wh, bh)
  if epoch % 100 == 0:
    print(f"Epoch: {epoch}")
    print(f"Loss: {funcion_loss}")
    print(f"Predicción: {output}")
    print(f"Pesos: {wo}")
    print(f"Bias: {bo}")
    print(f"Pesos: {wh}")
    print(f"Bias: {bh}")
    print()

#
```
